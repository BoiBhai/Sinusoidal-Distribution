```python
import torch
import torch.nn as nn

class SinCDFActivation(nn.Module):
    def __init__(self, n_points=100):
        super().__init__()
        self.s_raw = nn.Parameter(torch.tensor(0.0))
        self.k_raw = nn.Parameter(torch.tensor(0.0))
        self.n_points = n_points

    def forward(self, x):
        s = torch.exp(self.s_raw)
        k = torch.exp(self.k_raw)

        x = torch.clamp(x, 0, 1)

        # grid
        t = torch.linspace(0, 1, self.n_points, device=x.device)

        # denominator A(s,k)
        eps = 1e-6
        base = (torch.sin(torch.pi * t**s) + eps)**k
        A = torch.trapz(base, t)

        # numerator (scaled grid)
        t_scaled = x.unsqueeze(-1) * t
        integrand = (torch.sin(torch.pi * t_scaled**s) + eps)**k
        num = torch.trapz(integrand, t, dim=-1)

        return num / A
```
